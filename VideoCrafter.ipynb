# 🎬 VideoCrafter — type a topic, get a video. 100% free.

**How to use:**
1. Run **Step 1** below (click the ▶ button, takes ~1 minute, only needed once)
2. In **Step 2**, type your topic in the box on the right, then run it
3. Watch your video right here, then run **Step 3** to download it


```python
#@title Step 1: Setup (run once) { display-mode: "form" }
import os
if not os.path.exists('test-project'):
    !git clone -q -b claude/video-crafter-topic-generation-83cvfr https://github.com/khursheedalibht9-beep/test-project.git
%cd -q /content/test-project
!pip install -q -r requirements.txt
!apt-get -qq install -y espeak-ng > /dev/null
print('Setup done! Now go to Step 2 and type your topic.')
```

```python
#@title Step 2: Type your topic and run! { display-mode: "form" }
topic = "black holes"  #@param {type:"string"}
video_format = "shorts"  #@param ["shorts", "landscape", "square"]
voice = "male"  #@param ["male", "female", "uk-male", "uk-female", "hindi-male", "hindi-female", "urdu-male", "urdu-female"]
scenes = 6  #@param {type:"slider", min:3, max:10, step:1}
fast_preview = True  #@param {type:"boolean"}
#@markdown Optional, for real stock footage instead of gradient backgrounds — free key from [pexels.com/api](https://www.pexels.com/api/):
pexels_api_key = ""  #@param {type:"string"}

import os, subprocess, sys
if pexels_api_key.strip():
    os.environ['PEXELS_API_KEY'] = pexels_api_key.strip()
cmd = [sys.executable, '-m', 'videocrafter', topic,
       '--format', video_format, '--voice', voice,
       '--scenes', str(scenes), '--output', 'my_video.mp4']
if fast_preview:
    cmd.append('--preview')
subprocess.run(cmd, check=True)

from IPython.display import Video, display
display(Video('my_video.mp4', embed=True, width=320))
```

```python
#@title Step 3: Download your video { display-mode: "form" }
from google.colab import files
files.download('my_video.mp4')
```
